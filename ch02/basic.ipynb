```python
# concatenate

X = torch.arange(12, dtype=torch.float32).reshape(3, 4)
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
torch.cat((X, Y), dim = 0), torch.cat((X, Y), dim = 1)
```

```python
X == Y
```

```python
X.sum()
```

```python
# 节省内存
before = id(Y)
Y = X + Y
before, id(Y), before == id(Y)
```

```python
Z = torch.zeros_like(Y)
print('id(Z): ', id(Z))
Z[:] = X + Y
print('id(Z)', id(Z))
```

```python
before = id(X)
X += Y # X[:] = X + Y 也行
id(X) == before
```
