```python
# 平均值
A.mean(), A.sum() / A.numel()
```

```python
A.mean(axis=0) # 指定轴也可以降低维度，等价于
A.sum(axis=0) / A.shape[0], A.mean(axis=0) # A.shape[0], 等价于第零轴长度
```

```python
# 非降维求和，行第一维长度变1了
sum_A = A.sum(axis=0, keepdim=True)
sum_A
```

```python
print(A, sum_A, sep='\n')
A / sum_A
```

```python
sum_A = A.sum(axis=1, keepdim=True)
sum_A
```

```python
print(A, sum_A, sep='\n')
A / sum_A
```

```python
print(A.cumsum(axis=0))
print(A.cumsum(axis=1))
```
